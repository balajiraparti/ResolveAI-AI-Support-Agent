```python
def parse_response_ids(value):
    if pd.isna(value):
        return []
    
    return [
        int(x.strip())
        for x in str(value).split(",")
        if x.strip().isdigit()
    ]

df["response_ids"] = df["response_tweet_id"].apply(parse_response_ids)
```

```python
def find_parent_candidates(tweet_id, df):
    """
    Find tweets whose response_ids contain tweet_id.

    Parameters
    ----------
    tweet_id : int
        Tweet ID for which we want to find the parent.
    df : pd.DataFrame
        Full dataset containing tweet_id and response_ids.

    Returns
    -------
    pd.DataFrame
        Candidate parent tweets.
    """

    mask = df["response_ids"].apply(
        lambda ids: tweet_id in ids if isinstance(ids, list) else False
    )

    return df.loc[mask]
```

```python
def find_parent_from_response_ids(tweet_id, df):
    """
    Find tweets that explicitly list tweet_id in their response_ids.
    Returns candidate parent tweets.
    """

    mask = df["response_ids"].apply(
        lambda ids: tweet_id in ids if isinstance(ids, list) else False
    )

    candidates = df.loc[mask].copy()

    return candidates
```

```python
import pandas as pd

# ============================================================
# 1. Parent tweets
#    spotify_tweets.in_response_to_tweet_id -> df.tweet_id
# ============================================================

spotify_parent_matches = (
    spotify_tweets[
        spotify_tweets["in_response_to_tweet_id"].notna()
    ][
        [
            "tweet_id",
            "in_response_to_tweet_id",
            "text",
            "response_ids"          # maintain response_ids
        ]
    ]
    .merge(
        df[df["inbound"] == True][
            [
                "tweet_id",
                "author_id",
                "inbound",
                "created_at",
                "text"
            ]
        ],
        left_on="in_response_to_tweet_id",
        right_on="tweet_id",
        how="inner",
        suffixes=("_spotify", "_parent")
    )
)


# ============================================================
# 2. Response tweets
#    Maintain ORIGINAL response_ids
#    Use temporary column for matching
# ============================================================

spotify_response_matches = (
    spotify_tweets[
        [
            "tweet_id",
            "response_ids",
            "text"
        ]
    ]
    .assign(
        response_id_match=lambda x: x["response_ids"]
    )
    .explode("response_id_match")
    .merge(
        df[df["inbound"] == False][
            [
                "tweet_id",
                "author_id",
                "inbound",
                "created_at",
                "text",
                "in_response_to_tweet_id"
            ]
        ],
        left_on="response_id_match",
        right_on="tweet_id",
        how="inner",
        suffixes=("_spotify", "_response")
    )
)


# ============================================================
# Display results
# ============================================================

print(
    "Spotify → Customer parent matches:",
    len(spotify_parent_matches)
)

print(
    "Spotify → Customer response matches:",
    len(spotify_response_matches)
)


# ============================================================
# Parent matches
# ============================================================

display(
    spotify_parent_matches[
        [
            "tweet_id_spotify",
            "text_spotify",
            "in_response_to_tweet_id",
            "tweet_id_parent",
            "author_id",
            "inbound",
            "created_at",
            "text_parent",
            "response_ids"
        ]
    ]
)


# ============================================================
# Response matches
# ============================================================

display(
    spotify_response_matches[
        [
            "tweet_id_spotify",
            "text_spotify",

            # ORIGINAL COMPLETE response_ids list
            "response_ids",

            # Individual ID used for matching
            "response_id_match",

            "tweet_id_response",
            "author_id",
            "inbound",
            "created_at",
            "text_response",
            "in_response_to_tweet_id"
        ]
    ]
)
```

```python
# Spotify tweet IDs found in parent matches
parent_tweet_ids = set(
    spotify_parent_matches["tweet_id_spotify"]
    .dropna()
)

# Spotify tweet IDs found in response matches
response_tweet_ids = set(
    spotify_response_matches["tweet_id_spotify"]
    .dropna()
)

# Add match status
spotify_mentions["in_parent_matches"] = (
    spotify_mentions["tweet_id"].isin(parent_tweet_ids)
)

spotify_mentions["in_response_matches"] = (
    spotify_mentions["tweet_id"].isin(response_tweet_ids)
)

# ============================================================
# Summary
# ============================================================

print(
    "Total @SpotifyCares tweets:",
    len(spotify_mentions)
)

print(
    "Present in parent matches:",
    spotify_mentions["in_parent_matches"].sum()
)

print(
    "Present in response matches:",
    spotify_mentions["in_response_matches"].sum()
)

print(
    "Present in either parent or response:",
    (
        spotify_mentions["in_parent_matches"]
        | spotify_mentions["in_response_matches"]
    ).sum()
)

print(
    "Present in neither:",
    (
        ~spotify_mentions["in_parent_matches"]
        & ~spotify_mentions["in_response_matches"]
    ).sum()
)

# ============================================================
# Display
# ============================================================

display(
    spotify_mentions[
        [
            "tweet_id",
            "author_id",
            "created_at",
            "text",
            "in_parent_matches",
            "in_response_matches"
        ]
    ]
)
```
