```python
def parse_response_ids(value):
    if pd.isna(value):
        return []
    
    return [
        int(x.strip())
        for x in str(value).split(",")
        if x.strip().isdigit()
    ]

df["response_ids"] = df["response_tweet_id"].apply(parse_response_ids)
```

```python
def find_missing_responses(response_ids):
    return [
        x for x in response_ids
        if x not in tweet_ids
    ]

df["missing_response_ids"] = df["response_ids"].apply(
    find_missing_responses
)

print(
    "Rows with missing response tweets:",
    (df["missing_response_ids"].str.len() > 0).sum()
)
```

```python
def find_parent_candidates(tweet_id, df):
    """
    Find tweets whose response_ids contain tweet_id.

    Parameters
    ----------
    tweet_id : int
        Tweet ID for which we want to find the parent.
    df : pd.DataFrame
        Full dataset containing tweet_id and response_ids.

    Returns
    -------
    pd.DataFrame
        Candidate parent tweets.
    """

    mask = df["response_ids"].apply(
        lambda ids: tweet_id in ids if isinstance(ids, list) else False
    )

    return df.loc[mask]
```

```python
def find_referencing_tweets(tweet_id, df):
    mask = df["response_ids"].apply(
        lambda ids: tweet_id in ids if isinstance(ids, list) else False
    )

    return df.loc[mask]
```

```python
def find_parent_from_response_ids(tweet_id, df):
    """
    Find tweets that explicitly list tweet_id in their response_ids.
    Returns candidate parent tweets.
    """

    mask = df["response_ids"].apply(
        lambda ids: tweet_id in ids if isinstance(ids, list) else False
    )

    candidates = df.loc[mask].copy()

    return candidates
```
