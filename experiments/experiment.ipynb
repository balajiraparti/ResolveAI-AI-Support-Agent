```python
def find_parent_candidates(tweet_id, df):
    """
    Find tweets whose response_ids contain tweet_id.

    Parameters
    ----------
    tweet_id : int
        Tweet ID for which we want to find the parent.
    df : pd.DataFrame
        Full dataset containing tweet_id and response_ids.

    Returns
    -------
    pd.DataFrame
        Candidate parent tweets.
    """

    mask = df["response_ids"].apply(
        lambda ids: tweet_id in ids if isinstance(ids, list) else False
    )

    return df.loc[mask]
```

```python
def find_parent_from_response_ids(tweet_id, df):
    """
    Find tweets that explicitly list tweet_id in their response_ids.
    Returns candidate parent tweets.
    """

    mask = df["response_ids"].apply(
        lambda ids: tweet_id in ids if isinstance(ids, list) else False
    )

    candidates = df.loc[mask].copy()

    return candidates
```

```python
import re
import pandas as pd

def extract_author_id(text):
    """
    Extract an author ID/handle from the beginning of a tweet.

    Examples:
        '@SpotifyCares My account is not working'
            -> 'SpotifyCares'

        '@123456789 My payment failed'
            -> '123456789'

        'My account is not working'
            -> None
    """
    if pd.isna(text):
        return None

    text = str(text).strip()

    match = re.match(r"^@([A-Za-z0-9_]+)(?:\s+|$)", text)

    if match:
        return match.group(1)

    return None


spotify_tweets["extracted_author_id"] = (
    spotify_tweets["text"]
    .apply(extract_author_id)
)

spotify_tweets[
    ["tweet_id", "author_id", "text", "extracted_author_id"]
].head(20)
```
